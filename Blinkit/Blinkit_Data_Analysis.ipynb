## The column Item weight may have 1463 null values but we don't deal with it because the column is not so important
---

```python
df = df.drop('Item Weight',axis = 1)
```

```python
def col_unique_values(df):
    for column in df.columns:
        print("Column Name :",column)
        print("Unique Values :",df[column].unique())
        print('-' * 50)
```

```python
col_unique_values(df)
```

## let's focus on column : "Item Fat Content". It must be having only two unique values 'Low Fat' and 'Regular'
## So, Doing it :

```python
df['Item Fat Content'] = df['Item Fat Content'].replace({
    'low fat': 'Low Fat',
    'LF': 'Low Fat',
    'reg' : 'Regular'
})
    
```

```python
df['Item Fat Content'].unique()
```
