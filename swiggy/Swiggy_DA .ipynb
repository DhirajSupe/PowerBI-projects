## Let's find Unique values of some of the columns:
---

```python
for column in ['City', 'Cuisines', 'Food_Category','Payment_Method', 'Order_Status']:
    print(f"{column}:")
    print("Unique Count:", df[column].nunique())
    print("Unique Values:", df[column].unique())
    print('-' * 50)
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors = 'coerce')
```
